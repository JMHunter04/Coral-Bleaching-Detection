decoding the JSON coco RLE masks generated from CoralSCOP and applying to the pictures

```python
import json
import cv2
import numpy as np
from pycocotools import mask as maskUtils
```

```python
# loads the image
image = cv2.imread("/content/drive/MyDrive/dissertation/dataset/bleached_corals/8517429_68924ed843_o.jpg")

# loads the JSON mask
with open("/content/drive/MyDrive/dissertation/dataset/segmenteddataset/bleached_coral/8517429_68924ed843_o.json") as f:
    data = json.load(f)

# extracts rle from mask
rle = data["annotations"][0]["segmentation"]

# decodes the mask
mask = maskUtils.decode(rle)
```

```python
# combines mask with image
masked_image = image * mask[:, :, None]

# outputs image for check
cv2.imwrite("/content/drive/MyDrive/dissertation/dataset/maskeddataset/masked_image.jpg", masked_image)
```
