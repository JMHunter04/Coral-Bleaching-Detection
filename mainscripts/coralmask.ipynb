decoding the JSON coco RLE masks generated from CoralSCOP and applying to the pictures

```python
# libaries imported
import os
import json
import cv2
import numpy as np
from pycocotools import mask as maskUtils

# # #
# original dataset
image_root = "/content/drive/MyDrive/dissertation/dataset/originaldataset"
# JSON (mask) dataset
mask_root = "/content/drive/MyDrive/dissertation/dataset/segmenteddataset"
# output dataset
output_root = "/content/drive/MyDrive/dissertation/dataset/maskeddataset"

os.makedirs(output_root, exist_ok=True)
# two seperate folders for the two classes
classes = ["bleached_corals", "healthy_corals"]
# for loop cycling through both classes
for class_name in classes:
    # variables containing directories for each class
    # base image file path
    image_dir = os.path.join(image_root, class_name)
    # mask (JSON) file path
    mask_dir = os.path.join(mask_root, class_name)
    # output file path
    output_dir = os.path.join(output_root, class_name)

    os.makedirs(output_dir, exist_ok=True)
    # for each image in the current folder
    for filename in os.listdir(image_dir):
        # path to base image
        image_path = os.path.join(image_dir, filename)
        # finds JSON file with same name as base image
        json_name = os.path.splitext(filename)[0] + ".json"
        # creates JSON path
        json_path = os.path.join(mask_dir, json_name)
        # open current base image
        image = cv2.imread(image_path)

        with open(json_path) as f:
            data = json.load(f)

        # create empty mask
        height = data["image"]["height"]
        width = data["image"]["width"]
        mask = np.zeros((height, width), dtype=np.uint8)

        # combine all annotations into empty mask
        for ann in data["annotations"]:
            decoded = maskUtils.decode(ann["segmentation"])
            mask = np.logical_or(mask, decoded)

        mask = mask.astype(np.uint8)

        # apply mask to the base image
        masked_image = image * mask[:, :, None]

        # save new masked output image into output file path
        output_path = os.path.join(output_dir, filename)
        cv2.imwrite(output_path, masked_image)
        # confirmation output message for checking
        print(f"{class_name}/{filename}")

```
