```python
!pip install split_folders
import tensorflow as tf
import sklearn
import splitfolders
import os
from tensorflow.keras import datasets, layers, models
import numpy as np
import matplotlib.pyplot as plt
```

```python
input_folder = "/content/drive/MyDrive/dissertation/dataset/maskeddataset" #Enter Input Folder
```

```python
for class_dir in os.listdir(input_folder):
    class_path = os.path.join(input_folder, class_dir)
    if not os.path.isdir(class_path):
        continue

    for filename in os.listdir(class_path):
        if filename.startswith('.') or not filename.lower().endswith(('.jpg', '.jpeg', '.png')):
            file_path = os.path.join(class_path, filename)
            print(f"Removing: {file_path}")
            os.remove(file_path)
```

```python
output = "/content/drive/MyDrive/dissertation/dataset/splitdataset" #Enter Output Folder

splitfolders.ratio(input_folder, output=output, seed=2, ratio=(0.8,0.1,0.1), group_prefix=None, group=None,formats=None, move=False, shuffle=True)
```
